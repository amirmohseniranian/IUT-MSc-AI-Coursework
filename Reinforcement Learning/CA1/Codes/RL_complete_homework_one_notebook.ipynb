```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

GAMMA = 0.9
THETA = 0.001

# Common 5x5 gridworld utilities

states = [(r,c) for r in range(5) for c in range(5)]
actions = {
    0: "^",
    1: ">",
    2: "v",
    3: "<"
}

moves = {
    0: (-1,0),
    1: (0,1),
    2: (1,0),
    3: (0,-1)
}

def show_policy(policy):
    arrows = np.empty((5,5), dtype=str)
    for r,c in states:
        arrows[r,c] = actions[int(policy[r,c])]
    return pd.DataFrame(arrows)

def save_result(name, array):
    pd.DataFrame(array).to_csv(name + ".csv", index=False)
```

## Part 1 - Figure 5.3 Gridworld


```python

def part1_transition(state, action):
    r,c = state
    if state == (0,1):
        return (4,1),10
    if state == (0,3):
        return (2,3),5

    dr,dc = moves[action]
    nr,nc=r+dr,c+dc

    if nr<0 or nr>4 or nc<0 or nc>4:
        return state,-1

    return (nr,nc),0


def solve_part1():
    V=np.zeros((5,5))
    while True:
        old=V.copy()
        for s in states:
            value=0
            for a in range(4):
                ns,reward=part1_transition(s,a)
                value += 0.25*(reward+GAMMA*old[ns])
            V[s]=value

        if np.max(abs(V-old)) < THETA:
            break
    return V

V_part1=solve_part1()
print(np.round(V_part1,3))
save_result("part1_gridworld_value",V_part1)
```

## Part 3 - Policy Evaluation


```python

def transition(state, action):
    r,c=state
    dr,dc=moves[action]
    nr,nc=r+dr,c+dc

    if nr<0 or nr>4 or nc<0 or nc>4:
        return state,-1

    return (nr,nc),0


def policy_evaluation():
    V=np.zeros((5,5))

    while True:
        delta=0

        for s in states:
            old=V[s]
            value=0

            for a in range(4):
                ns,reward=transition(s,a)
                value += .25*(reward+GAMMA*V[ns])

            V[s]=value
            delta=max(delta,abs(old-value))

        if delta < THETA:
            break

    return V


V_policy_eval=policy_evaluation()
print(np.round(V_policy_eval,3))
save_result("policy_evaluation",V_policy_eval)
```

## Part 4 - Policy Iteration


```python

def policy_iteration():

    V=np.zeros((5,5))
    policy=np.zeros((5,5),dtype=int)

    stable=False

    while not stable:

        while True:
            delta=0

            for s in states:
                a=policy[s]
                ns,reward=transition(s,a)
                old=V[s]
                V[s]=reward+GAMMA*V[ns]
                delta=max(delta,abs(old-V[s]))

            if delta<THETA:
                break

        stable=True

        for s in states:
            old_action=policy[s]

            values=[]

            for a in range(4):
                ns,reward=transition(s,a)
                values.append(reward+GAMMA*V[ns])

            policy[s]=np.argmax(values)

            if old_action != policy[s]:
                stable=False

    return V,policy


V_pi, policy_pi = policy_iteration()

print(np.round(V_pi,3))
print(show_policy(policy_pi))

save_result("policy_iteration_value",V_pi)
save_result("policy_iteration_policy",policy_pi)
```

## Part 5 - Value Iteration


```python

def value_iteration():

    V=np.zeros((5,5))

    while True:
        delta=0

        for s in states:

            values=[]

            for a in range(4):
                ns,reward=transition(s,a)
                values.append(reward+GAMMA*V[ns])

            old=V[s]
            V[s]=max(values)

            delta=max(delta,abs(old-V[s]))

        if delta<THETA:
            break

    policy=np.zeros((5,5),dtype=int)

    for s in states:

        values=[]

        for a in range(4):
            ns,reward=transition(s,a)
            values.append(reward+GAMMA*V[ns])

        policy[s]=np.argmax(values)

    return V,policy


V_vi, policy_vi=value_iteration()

print(np.round(V_vi,3))
print(show_policy(policy_vi))

save_result("value_iteration_value",V_vi)
save_result("value_iteration_policy",policy_vi)
```
